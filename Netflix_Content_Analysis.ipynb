```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6,4))

df["type"].value_counts().plot(
    kind="bar",
    color=["red", "black"]
)

plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("visualization1.png")

plt.show()
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

content_per_year = (
    df["date_added"]
    .dt.year
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(10,5))

content_per_year.plot(
    kind="line",
    marker="o"
)

plt.title("Netflix Content Added by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.tight_layout()

plt.savefig("visualization2.png")

plt.show()
```

```python
top_countries = (
    df["country"]
    .dropna()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,5))

top_countries.plot(kind="bar")

plt.title("Top 10 Countries by Number of Titles")
plt.xlabel("Country")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("visualization3.png")

plt.show()
```

```python
top_ratings = (
    df["rating"]
    .dropna()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(8,5))

top_ratings.plot(kind="bar")

plt.title("Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("visualization4.png")

plt.show()
```

```python
print("Total records:", len(df))

print("\nContent Type Distribution:")
print(df["type"].value_counts())

print("\nTop Countries:")
print(df["country"].value_counts().head())

print("\nTop Ratings:")
print(df["rating"].value_counts().head())
```

# Conclusion
## Key Findings
1. Movies represent the majority of Netflix content.
2. Netflix experienced rapid content growth between 2016 and 2019.
3. The United States is the largest contributor of content.
4. TV-MA and TV-14 are the most common content ratings.

## Limitation
The dataset represents a snapshot of Netflix content and may not reflect future catalog updates.
